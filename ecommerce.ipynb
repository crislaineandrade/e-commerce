```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
#visualizing data
ecommerce = pd.read_csv('Ecommerce_DBS.csv')

display(ecommerce.head(5))
display(ecommerce.info())
display(ecommerce.describe())
```

```python
#data treatments

ecommerce.columns = ecommerce.columns.str.lower()
ecommerce.columns = ecommerce.columns.str.replace(' ', '_')
ecommerce['purchase_date'] = pd.to_datetime(ecommerce['purchase_date'], format='%d/%m/%Y')
```

```python
#which product sells the most?

products_bought_most = ecommerce.groupby('product_category')['quantity'].sum().reset_index().sort_values(ascending=False, by='quantity')
display(products_bought_most)

#visualize
products_bought_most.plot(kind='bar', x='product_category', y='quantity', title='Product Categories X Quantity', xlabel='Product Category', ylabel='Quantity')
```
