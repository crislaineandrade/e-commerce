```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
#visualizing data
ecommerce = pd.read_csv('Ecommerce_DBS.csv')

display(ecommerce.head(5))
display(ecommerce.info())
display(ecommerce.describe())
```

```python
#data treatments

ecommerce.columns = ecommerce.columns.str.lower()
ecommerce.columns = ecommerce.columns.str.strip()
ecommerce.columns = ecommerce.columns.str.replace(' ', '_')
ecommerce['purchase_date'] = pd.to_datetime(ecommerce['purchase_date'], format='%d/%m/%Y')
```

```python
#which product sells the most?

products_bought_most = ecommerce.groupby('product_category')['quantity'].sum().reset_index().sort_values(ascending=False, by='quantity')
display(products_bought_most)

#visualize
products_bought_most.plot(kind='bar', x='product_category', y='quantity', title='Product Categories X Quantity', xlabel='Product Category', ylabel='Quantity')
```

```python
#Where does the largest source of revenue come from

sources_group = ecommerce.groupby('source')['total_purchase_amount'].sum().reset_index().sort_values(by='total_purchase_amount', ascending=False)


sources_group['revenue_share'] = sources_group['total_purchase_amount'] / sources_group['total_purchase_amount'].sum() * 100

purchase_by_source = (ecommerce.groupby('source').size().reset_index(name='number_of_purchases').sort_values('number_of_purchases', ascending=False))


sources_group = sources_group.merge(purchase_by_source, on='source')


display(sources_group)
```

```python
#highest revenue source by customer age


age_purchase_revenue = (
    ecommerce
    .groupby(['customer_age', 'source'])['total_purchase_amount']
    .sum()
    .reset_index()
)


age_purchase_revenue = age_purchase_revenue.loc[age_purchase_revenue.groupby('customer_age')['total_purchase_amount'].idxmax()].sort_values('customer_age')

plt.figure(figsize=(14,7))
sns.barplot(
    data=age_purchase_revenue,
    x='customer_age',
    y='total_purchase_amount',    
    hue='source'
)

plt.xlabel('Customer Age')
plt.ylabel('Total Purchase Amount')

plt.title('Highest Revenue Source by Customer Age')
plt.xticks(rotation=60)

plt.show()
```
